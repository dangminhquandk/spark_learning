```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

from pyspark.sql import SparkSession
from graphframes import GraphFrame
```

```python
# Initialize
spark = (
    SparkSession.builder
    .appName("Macbook_Air_M3_Safe_GraphFrames")
    .master("local[4]")
    .config("spark.driver.memory", "4g")
    .config("spark.driver.memoryOverhead", "512m")
    .config("spark.sql.shuffle.partitions", "4")
    .config("spark.default.parallelism", "4")
    .config("spark.jars.packages", "io.graphframes:graphframes-spark4_2.13:0.12.2")
    .getOrCreate()
)

# Put Checkpoint directory for loop algorithms
spark.sparkContext.setCheckpointDir("/tmp/spark-checkpoints")
```

```python
# Create Vertices
vertices_data = [
    ("ACC_101", "Nguyen Van A", "Personal", 15000.0, 0),
    ("ACC_102", "Tran Thi B", "Personal", 50000.0, 0), 
    ("ACC_103", "Le Van C", "Business", 200000.0, 1), # Nhãn 1: Tài khoản gian lận nghi ngờ
    ("ACC_104", "Pham Van D", "Personal", 8000.0, 1), # Nhãn 1: Tài khoản gian lận nghi ngờ 
    ("ACC_105", "Hoang Thi E", "Personal", 12000.0, 1), # Nhãn 1: Tài khoản gian lận nghi ngờ 
    ("ACC_106", "Cty TNHH X", "Business", 500000.0, 0), 
    ("ACC_107", "Vu Van F", "Personal", 3000.0, 0)
]

vertices = spark.createDataFrame(
    vertices_data,
    ["id", "name", "account_type", "balance", "label"]
)
```

```python
# Create Edges
edges_data = [
    ("ACC_101", "ACC_102", 2000.0, "2026-09-01 08:00:00"),
    ("ACC_102", "ACC_106", 15000.0, "2026-09-01 09:30:00"), 
    ("ACC_107", "ACC_101", 1000.0, "2026-09-01 10:15:00"), 
    
    # Vòng tròn giao dịch gian lận/rửa tiền khép kín (103 -&gt; 104 -&gt; 105 -&gt; 103) 
    ("ACC_103", "ACC_104", 45000.0, "2026-09-02 14:00:00"), 
    ("ACC_104", "ACC_105", 44500.0, "2026-09-02 14:05:00"), 
    ("ACC_105", "ACC_103", 44000.0, "2026-09-02 14:10:00"), 
    
    # Giao dịch chuyển tiền từ vòng gian lận ra tài khoản bên ngoài 
    ("ACC_105", "ACC_101", 5000.0, "2026-09-02 15:00:00")
]

edges = spark.createDataFrame(
    edges_data,
    ["src", "dst", "amount", "timestamp"]
)
```

```python
# Initialize GraphFrame model
g = GraphFrame(vertices, edges)

print("=== Vertices ===")
g.vertices.show()

print("=== Edges ===")
g.edges.show()
```

```python
# II. Detect cycle 
from pyspark.sql.functions import col

# Declare: A -> B -> C
raw_cycles = g.find("(a)-[e1]->(b); (b)-[e2]->(c); (c)-[e3]->(a)")

# Filter for removing duplicated cases
fraud_cycles = raw_cycles.filter("a.id != b.id AND b.id != c.id AND  a.id != c.id")

# Extract
print("=== Cycle: A -> B -> C -> A ===")
fraud_cycles.select(
    col("a.id").alias("Acc_A"),
    col("e1.amount").alias("Tien_A_ck_B"),
    col("b.id").alias("Acc_B"),
    col("e2.amount").alias("Tien_B_ck_C"),
    col("c.id").alias("Acc_C"),
    col("e3.amount").alias("Tien_C_ck_A")
).show(truncate = False)


# Compute
print("=== TỔNG SỐ TIỀN BỊ XOAY VÒNG ===")
fraud_cycles.withColumn("Tong_tien_xoay_vong", col("e1.amount") + col("e2.amount") + col("e3.amount")) \
                        .select("a.id", "b.id", "c.id", "Tong_tien_xoay_vong") \
                        .show()
```

```python
# CONNECTED COMPONENTS & PAGERANK
from pyspark.sql.functions import col, desc

# 1. Cấu hình bắt buộc cho thuật toán lặp của GraphFrame
spark.sparkContext.setCheckpointDir("/tmp/spark-checkpoints")

# 2. Khởi chạy Connected Components
cc_df = g.connectedComponents()

print("=== Overview Components ===")
# Thống kê số lượng tài khoản trong mỗi cụm
cc_summary = (
    cc_df.groupBy("component")
    .count()
    .withColumnRenamed("count", "so_luong_tai_khoan")
)
cc_summary.show()

print("=== CHI TIẾT TÀI KHOẢN VÀ MÃ CỤM (COMPONENT ID) ===")
# Lưu ý kiểm tra lại các cột có thực sự tồn tại trong vertices ban đầu hay không
cc_df.select("id", "component").show()
# Hoặc nếu bảng đỉnh có đầy đủ thuộc tính:
# cc_df.select("id", "name", "account_type", "balance", "label", "component").orderBy("component", "id").show()

# Bỏ hoàn toàn tham số weightCol
pr_result = g.pageRank(resetProbability=0.15, maxIter=10)

print("=== 2. BẢNG XẾP HẠNG ĐỘ TRUNG TÂM / DÒNG TIỀN (PAGERANK) ===")
# pr_result.vertices sẽ tự động chứa thêm cột 'pagerank'
pr_result.vertices.orderBy(desc("pagerank")).show(truncate=False)
```

```python
# ==============================================================================
# BƯỚC 4: TRÍCH XUẤT GRAPH FEATURES & HUẤN LUYỆN MÔ HÌNH MACHINE LEARNING
# ==============================================================================
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.sql.functions import coalesce, col, lit

# ------------------------------------------------------------------------------
# 1. GOM TẤT CẢ GRAPH FEATURES THÀNH MỘT BẢNG DỮ LIỆU TẬP TRUNG
# ------------------------------------------------------------------------------
# Lấy số lượng giao dịch đi vào (inDegrees) và đi ra (outDegrees)
in_deg = g.inDegrees.withColumnRenamed("inDegree", "in_degree")
out_deg = g.outDegrees.withColumnRenamed("outDegree", "out_degree")

# Rút trích bảng PageRank và Component đã tính ở Bước 3
pr_df = pr_result.vertices.select("id", "pagerank")
cc_sub = cc_df.select("id", "component")

# JOIN tất cả đặc trưng truyền thống + đặc trưng đồ thị vào bảng Vertices ban đầu
full_features_df = (
    g.vertices.join(pr_df, "id", "left")
    .join(cc_sub, "id", "left")
    .join(in_deg, "id", "left")
    .join(out_deg, "id", "left")
    .na.fill(
        0, ["in_degree", "out_degree"]
    )  # Điền giá trị 0 cho các tài khoản không có giao dịch
)

print("=== BẢNG DỮ LIỆU ĐÃ TÍCH HỢP GRAPH FEATURES ===")
full_features_df.select(
    "id",
    "name",
    "balance",
    "pagerank",
    "component",
    "in_degree",
    "out_degree",
    "label",
).show()

# ------------------------------------------------------------------------------
# 2. CHUẨN BỊ VECTOR ĐẶC TRƯNG (FEATURE PIPELINE) CHO PYSPARK MLLIB
# ------------------------------------------------------------------------------
# Mã hóa cột dạng chữ 'account_type' ('Personal', 'Business') thành dạng số
indexer = StringIndexer(inputCol="account_type", outputCol="account_type_idx")
indexed_df = indexer.fit(full_features_df).transform(full_features_df)

# Gom tất cả các cột đặc trưng (cả thuộc tính tài khoản + Graph Features) thành 1 Vector
feature_cols = [
    "balance",
    "account_type_idx",
    "pagerank",
    "component",
    "in_degree",
    "out_degree",
]
assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
ml_dataset = assembler.transform(indexed_df)

print("=== MẪU DỮ LIỆU ĐẦU VÀO MÔ HÌNH ML (VECTOR FEATURES) ===")
ml_dataset.select("id", "name", "features", "label").show(truncate=False)

# ------------------------------------------------------------------------------
# 3. HUẤN LUYỆN MÔ HÌNH PHÂN LOẠI GIAN LẬN (RANDOM FOREST CLASSIFIER)
# ------------------------------------------------------------------------------
rf = RandomForestClassifier(
    labelCol="label", featuresCol="features", numTrees=10, maxDepth=5, seed=42
)

model = rf.fit(ml_dataset)

# Dự đoán xác suất gian lận cho toàn bộ danh sách tài khoản
predictions = model.transform(ml_dataset)

print("=== KẾT QUẢ DỰ ĐOÁN TÀI KHOẢN GIAN LẬN (PREDICTIONS) ===")
predictions.select(
    "id",
    "name",
    "label",
    col("prediction").cast("int").alias("du_doan_gian_lan"),
    "probability",
).show(truncate=False)
```

```python
from pyspark.ml.functions import vector_to_array
from pyspark.sql.functions import col, when

# 1. Chuyển VectorUDT sang Array rồi trích xuất phần tử index 1
decision_engine_df = (
    predictions.withColumn(
        "fraud_probability", vector_to_array(col("probability"))[1]
    )
    .withColumn(
        "muc_do_rui_ro",
        when(col("fraud_probability") >= 0.8, "CAO (HIGH RISK)")
        .when(
            (col("fraud_probability") >= 0.5)
            & (col("fraud_probability") < 0.8),
            "TRUNG BÌNH (MEDIUM RISK)",
        )
        .otherwise("THẤP (LOW RISK)"),
    )
    .withColumn(
        "hanh_dong_he_thong",
        when(
            col("fraud_probability") >= 0.8,
            "🛑 TỰ ĐỘNG ĐÓNG BĂNG TÀI KHOẢN & BLOCK GIAO DỊCH",
        )
        .when(
            (col("fraud_probability") >= 0.5)
            & (col("fraud_probability") < 0.8),
            "⚠️ CHUYỂN BÁO CÁO CHO ĐỘI AUDIT THẨM ĐỊNH THỦ CÔNG",
        )
        .otherwise("✅ CHO PHÉP GIAO DỊCH BÌNH THƯỜNG"),
    )
)

# 2. Xuất báo cáo
decision_engine_df.select(
    col("id").alias("Ma_TK"),
    col("name").alias("Ten_Chu_TK"),
    col("component").alias("Cum_Nhom"),
    col("fraud_probability").alias("Xac_Suat_Gian_Lan"),
    col("muc_do_rui_ro").alias("Rui_Ro"),
    col("hanh_dong_he_thong").alias("Quyet_Dinh_Xu_Ly"),
).orderBy(col("Xac_Suat_Gian_Lan").desc()).show(truncate=False)
```
