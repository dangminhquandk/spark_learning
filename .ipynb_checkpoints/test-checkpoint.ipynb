```python
from pyspark.sql import SparkSession
from graphframes import GraphFrame

spark = (
    SparkSession.builder
    .appName("Macbook_Air_M3_Safe_GraphFrames")
    .master("local[4]")
    .config("spark.driver.memory", "4g")
    .config("spark.driver.memoryOverhead", "512m")
    .config("spark.sql.shuffle.partitions", "4")
    .config("spark.default.parallelism", "4")
    .config("spark.jars.packages", "org.graphframes:graphframes_2.13:0.9.0-spark4.0-s_2.13")
    .getOrCreate()
)
```

```python
# 1. Create a Vertices DataFrame - GraphFrames requires an 'id' column
vertices = spark.createDataFrame([
    ("`", "Alice", 28),
    ("2", "Bob", 32),
    ("3", "Charlie", 25)
], ["id", "name", "age"])

# 2. Create a Edges DataFrame - GraphGrames
edges = spark.createDataFrame([
    ("1", "2", "friend"),
    ("2", "3", "follow"),
    ("3", "1", "friend")
], ["src", "dst", "relationship"])

# 3. Initialize GraphFrame
g  = GraphFrame(vertices, edges)

# 4.Test the vertex degree calculation
print("--- Quantity of Degrees ---")
g.degrees.show()
```
