# NULL VALUE

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Practice").getOrCreate()
```

```python
df=spark.read.csv("test2.csv",header=True,inferSchema=True)
```

```python
df.show()
```

## Drop column

```python
df.drop("Name").show()
```

```python
df.show()
```

## na.drop(): row drop, NaN, NULL

```python
df.na.drop().show()
```

```python
## how; default='any'
df.na.drop(how='all').show()
```

```python
df.na.drop(how='any').show()
```

```python
## threshold
df.na.drop(how='any',thresh=3).show()
```

```python
## subset
df.na.drop(how='any',subset=['Experience']).show()
```

```python
df.show()
```

```python
## subset
df.na.drop(how='any',subset=['age']).show()
```

## Filling the Missing value

```python
## null -> Missing Values
## subset = None
df.na.fill('Missing Values').show()
```

```python
## subset is not None (= Experience, age)
df.na.fill('Missing Values',['Experience','age']).show()
```

```python
df.show()
```

## Cần tính toán chứ ko chỉ điền giá trị cho sẵn

```python
from pyspark.ml.feature import Imputer

# 1. Khai báo danh sách cột cần xử lý vào một biến dùng chung
cols = ['age', 'Experience', 'Salary']

# 2. Khởi tạo công cụ Imputer
imputer = Imputer(
    inputCols=cols, 
    outputCols=[f"{c}_imputed" for c in cols]
).setStrategy("median")

# 3. Quét tính số trung vị và điền luôn vào bảng mới
df_imputed = imputer.fit(df).transform(df)

# 4. Hiển thị kiểm tra các cột trước và sau khi điền
df_imputed.show()
```

```python
spark.stop()
```
